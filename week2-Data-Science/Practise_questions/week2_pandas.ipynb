Dataset: SQL_Sales_Dataset_200_Rows.csv (200 orders, 10 columns)

## Q1: Load a CSV file using Pandas and display basic info

```python
import pandas as pd

df = pd.read_csv("SQL_Sales_Dataset_200_Rows.csv")
df.head()
```

```python
print(df.shape)
df.info()
```

```python
df.describe()
```

## Q2: Handle missing values and duplicates using Pandas

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.dropna()
df = df.drop_duplicates()
df.shape
```

No missing values or duplicates were found, so no rows were removed. The dataset has 200 rows and 10 columns.

## Q3: Group data by category and find total revenue

```python
revenue_by_category = df.groupby("category")["total_price"].sum().sort_values(ascending=False)
revenue_by_category
```

Furniture generates the highest total revenue, followed by Grocery, Electronics and Clothing.

## Q4: Sort data by multiple columns using Python

```python
df_sorted = df.sort_values(by=["category", "total_price"], ascending=[True, False])
df_sorted.head(10)
```

The data is sorted by category (A to Z) and, within each category, by total_price from highest to lowest.

## Q5: Create a correlation matrix for numerical columns

```python
corr = df[["quantity", "unit_price", "total_price"]].corr()
corr
```

total_price is positively correlated with both quantity (about 0.66) and unit_price (about 0.67). quantity and unit_price are almost uncorrelated.
